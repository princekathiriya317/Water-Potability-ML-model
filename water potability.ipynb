## Data Cleaning 

```python
df.isnull().sum()
```

```python
# observation 
# in this dataset total 3 columns have missing values. ph has 491 missing values ,sulfate has 781 missing values ,Trihalomethanes has 162 missing values
```

```python
df['ph']
```

```python
# Fill Missing values 
```

```python
# Fill missing values based on the median.
```

```python
df['ph'] = df['ph'].fillna(df['ph'].median())
```

```python
df['Sulfate'] = df['Sulfate'].fillna(df['Sulfate'].median())
```

```python
df['Trihalomethanes'] = df['Trihalomethanes'].fillna(df['Trihalomethanes'].median())
```

# Outlier Detection

```python
features = df.columns.drop("Potability")

plt.figure(figsize=(15, 10))

for i, col in enumerate(features, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[col], color="skyblue")
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
#  this boxplot is show a outliers in this dataset , all columns have a many outlier.
```

# Handle Outliers

```python
from scipy.stats import zscore
```

```python
z = np.abs(zscore(df))
```

```python
dfn = df[(z<3).all(axis=1)]
```

```python
dfn.shape , df.shape 
```

```python
# using zscore method remove outliers in the dataset 
```

```python
dfn.head()
```

# Data Preproccessing 

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
x = dfn.iloc[:,:-1]
```

```python
y = dfn.iloc[:,-1]
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=.20 , random_state=42)
```

```python
x_train.shape , x_test.shape ,y_train.shape ,y_test.shape
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# standard scaler used because in this dataset ffeatures values are equally by a machine learning model .
```

# Model Train and Test

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=.20 , random_state=42)
```

```python
x_train.shape , x_test.shape ,y_train.shape ,y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score , confusion_matrix ,classification_report
import warnings 
warnings.filterwarnings('ignore')
```

```python
lr = LogisticRegression(class_weight='balanced')
```

```python
lr.fit(x_train_scaled,y_train)
```

```python
y_pred = lr.predict(x_test_scaled)
```

```python
print("Accuracy Score:" , accuracy_score(y_test ,y_pred))
```

```python
print("Confussion Metrix:\n",confusion_matrix(y_test,y_pred))
```

```python
print("classification Report:\n",classification_report(y_test , y_pred))
```

```python
# logistic regression model is not give a best accuracy because water potability data is overlays . in this situation i used random forest classifier.
```

# RandomForestClassifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(n_estimators=100 , random_state=42)
```

```python
rf.fit(x_train_scaled , y_train)
```

```python
y_pred1 = rf.predict(x_test_scaled)
```

```python
print("Accuracy Score:" , accuracy_score(y_test ,y_pred1))
```

```python
accuracy = accuracy_score(y_test, y_pred1)
print(f"Random Forest Accuracy: {accuracy * 100:.2f}%")
```

```python
print("Confussion Metrix:\n",confusion_matrix(y_test,y_pred1))
```

```python
print("classification Report:\n",classification_report(y_test , y_pred1))
```
